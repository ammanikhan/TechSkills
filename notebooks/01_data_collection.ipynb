## Data Collection and Preparation

### Data Source
We collected job postings from Hacker News monthly "Who is Hiring?" threads from January 2023 to November 2026.

Our original dataset contained 15,252 posts.

### Data Cleaning
We used Python and Pandas to clean and organize the data. This included:

- Removing duplicate posts and very short comments.
- Removing HTML formatting.
- Extracting company names, job roles, and locations.
- Identifying technical skills using a list of 26 technologies.
- Categorizing work arrangements as remote, hybrid, onsite, or unknown.

After cleaning, we had 15,197 unique posts.

### Data Preparation
For our initial skill analysis, we filtered the dataset to include likely hiring posts containing at least one recognized technical skill.

This resulted in 9,124 posts.

We kept the full cleaned dataset separately for analyses that may require different filtering.

### Data Limitations
Some job postings have inconsistent formatting, which makes it difficult to accurately extract roles and locations.

Our skill extraction only identifies the 26 technologies included in our list, so other skills may be missed. Work arrangements are also estimated using keywords and may not always be accurate.

We will continue improving and validating the dataset as the project progresses.

```python
from pathlib import Path
import pandas as pd

DATA = Path("../data/processed/jobs_cleaned.csv")
if not DATA.exists():
    DATA = Path("data/processed/jobs_cleaned.csv")

cleaned_df = pd.read_csv(DATA)
print(cleaned_df.shape)
display(cleaned_df.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from pathlib import Path

data_path = Path("../data/processed/jobs_analysis.csv")
if not data_path.exists():
    data_path = Path("data/processed/jobs_analysis.csv")

analysis_df = pd.read_csv(data_path)

skills = Counter()

for value in analysis_df["skills"].dropna():
    for skill in value.split(", "):
        skills[skill] += 1

top_skills = pd.Series(dict(skills.most_common(10)))

plt.figure(figsize=(10, 5))
top_skills.sort_values().plot(kind="barh")

plt.title("Top 10 Technical Skills in Hacker News Job Posts")
plt.xlabel("Number of Posts")
plt.ylabel("Technical Skill")

plt.tight_layout()
plt.show()
```

### EDA Findings: Most Common Technical Skills

We analyzed 9,124 Hacker News hiring posts from 2023–2026 that contained at least one recognized technical skill.

React (3,356), Python (3,345), and TypeScript (3,000) were the most frequently mentioned skills, followed by AWS (2,096) and Go (1,524).

On average, each posting mentioned 2.73 of the 26 skills we tracked.

These results suggest that web development, programming, and cloud technologies are common in our dataset. However, our analysis only tracks selected skills, so other technologies may not be represented.
