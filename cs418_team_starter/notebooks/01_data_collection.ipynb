**Data:** Hacker News monthly “Who is Hiring” threads. **Question:** How have requested tech skills changed, and what can job postings tell us about AI and entry-level roles?

**Limitations:** These posts are not a representative sample of the entire labor market; skill keyword mentions do not prove a skill is required.


```python
from pathlib import Path
import pandas as pd
DATA = Path("../data/processed/jobs_cleaned.csv")
df = pd.read_csv(DATA)
print(df.shape)
display(df.head())
```

```python
display(df.isna().sum().rename("missing"))
print("Years:", sorted(df.year.dropna().unique()))
print("Duplicates:", df.id.duplicated().sum())
```

## Data collection and cleaning
Collected top-level comments from Hacker News monthly hiring threads via the public Algolia API. Removed HTML tags and short/duplicate comments. Extracted skills using a transparent keyword dictionary. Some structured fields are intentionally missing because postings are free-form. Validate a sample manually before drawing conclusions.
